```python
import pandas as pd
import numpy as np
projects_completed = [5, 8, 6, 10, 7]
print("DISCRETE DATA:",projects_completed)
salaries = [35000.5, 42000.0, 39000.8, 48000.2, 41000.6]
print("CONTINUOUS DATA:",salaries)
departments = ["HR", "IT", "Finance", "IT", "HR"]
print("NOMINAL DATA:",departments)
performance_ratings = ["Excellent", "Good", "Excellent", "Average", "Good"]
print("ORDINAL DATA:",performance_ratings)
employee_ages = [25, None, 30, 28, 26]
print("DISCRETE DATA:",employee_ages)
```

```python
fi=np.mean(projects_completed)
print("THE AVERAGE OF PROJECTS COMPLETED:",fi)
```

```python
di=np.mean(salaries)
print("THE AVERAGE OF SALARIES IS:",di)
```

```python
sw=np.max(salaries)
print("THE MAXIMUM SALARY IS:",sw)
```

```python
dw=np.min(salaries)
print("THE MINIMUM SALARY IS:",dw)
```

```python
df=pd.DataFrame({"employee_ages":employee_ages})
se=(df.isnull().sum())
print("MISSING EMPLOYEE DATA FOUND:",se)
```

```python
fe=set(departments)
print("THE UNIQUE DEPARTMENTS ARE:",fe)
```

```python
rating_order = {"Average": 1, "Good": 2, "Excellent": 3}
numeric_ratings = [rating_order[r] for r in performance_ratings]
print("PERFORMANCE  RATINGS:", numeric_ratings)
```
