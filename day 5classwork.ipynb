```python
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
salaries = [[25000], [40000], [55000], [70000], [85000]]
scaler=MinMaxScaler()
scaled_data=scaler.fit_transform(salaries)
normalised=scaler.fit_transform(salaries)
print("THE SCALED DATA IS:",scaled_data)
print("THE NORMALISED DATA IS :",normalised)
print("THE MIN AND MAX OF SCALED AND NORMALISED VALUES ARE:")
print(scaled_data.min())
print(scaled_data.max())
print(normalised.min())
print(normalised.max())
```
