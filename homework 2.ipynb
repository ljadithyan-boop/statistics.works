```python
users = [
   {"premium": True, "new_release": True}, 
   {"premium": True, "new_release": False},
   {"premium": True, "new_release": True},
   {"premium": False, "new_release": True},
   {"premium": False, "new_release": False},
   {"premium": False, "new_release": True},
   {"premium": False, "new_release": False},
   {"premium": True, "new_release": True}
]
population=20000
print("THE SAMPLE IS :",users)
print("THE POPULATION IS :",population)

A=[x for x in users if x["premium"]]
B=[x for x in users if x["new_release"]]
B_given_A=[x for x in users if x["premium"] and x["new_release"]]
P_A=len(A)/len(users)
P_B=len(B)/len(users)
P_B_given_A = len(B_given_A) / len(A)
P_A_given_B= (P_B_given_A * P_A) / P_B
print("THE BAYES THEORM APPLIED IS:",P_A_given_B)
```
