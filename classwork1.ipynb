```python
import pandas as pd
import numpy as np
student_counts = [25, 30, 28, 32, 27]
print("DISCRETE DATA:",student_counts)
temperatures = [36.5, 37.2, 36.8, 37.0, 36.9]
print("CONTINUOUS DATA:",temperatures)
favorite_subjects = ["Math", "Science", "Math", "English", "Science"]
print("NOMINAL DATA:",favorite_subjects)
performance_levels = ["High", "Medium", "High", "Low", "Medium"]
print("ORDINAL DATA:",performance_levels)
```

```python
import numpy as np
sep=np.mean(student_counts)
print("THE AVERAGE OF DISCRETE DATA:",sep)
```

```python
NEP=np.mean(temperatures)
print("THE AVERAGE OF THE TEMPERATURES IS :",NEP)
```

```python
se=np.min(temperatures)
print("THE MINIMUM TEMPERATURE IS:",se)
```

```python
fe=np.max(student_counts)
print("THE MAXIMUM STUDENT COUNT:",fe)
```

```python
fe=set(favorite_subjects)
print("THE UNIQUE SUBJECTS ARE:", fe)
```

```python
order = {"Low": 1, "Medium": 2, "High": 3}
numeric_ratings = [order[r] for r in performance_levels]
print("Ordered Ratings:", numeric_ratings)
```
