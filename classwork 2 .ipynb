```python
customers = [

   {"premium": True, "promo_email": True},

   {"premium": True, "promo_email": True},

   {"premium": True, "promo_email": False},

   {"premium": False, "promo_email": True},

   {"premium": False, "promo_email": False},

   {"premium": False, "promo_email": True}
]
population=5000
print("THE SAMPLE IS:",customers)
print("THE POPULATION IS:",population)
```

```python
A=[x for x in customers if x["premium"]]
B=[x for x in customers if x["promo_email"]]
B_given_A=[x for x in customers if x["premium"]and x["promo_email"]]
P_A = len(A) / len(customers)
P_B = len(B) / len(customers)
P_B_given_A = len(B_given_A) / len(A)
P_A_given_B = (P_B_given_A * P_A) / P_B
print(P_A_given_B)
```
